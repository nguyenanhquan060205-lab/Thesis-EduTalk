Notebook này thực hiện:
- Load và kiểm tra dữ liệu thô
- Thống kê mô tả (describe)
- Phân phối nhãn (ngành học)
- Visualize mối quan hệ điểm thi ↔ ngành
- Phát hiện missing values, outliers

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cấu hình hiển thị
pd.set_option('display.max_columns', None)
plt.rcParams['figure.figsize'] = (12, 6)
sns.set_theme(style='whitegrid')

print('✅ Libraries loaded')
```

## 1. Load dữ liệu thô

```python
# Load dữ liệu — thay đường dẫn tương ứng
df = pd.read_csv('../data/raw/huit_admissions_data.csv')

print(f'Shape: {df.shape}')
print(f'Columns: {df.columns.tolist()}')
df.head()
```

## 2. Thống kê mô tả

```python
# Tổng quan
df.info()
print('\n--- Missing values ---')
print(df.isnull().sum())
print('\n--- Describe ---')
df.describe()
```

## 3. Phân phối nhãn (ngành học)

```python
# TODO: thay 'major' bằng tên cột nhãn thật trong dataset
label_col = 'major'

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Bar chart
counts = df[label_col].value_counts()
counts.plot(kind='bar', ax=axes[0], color='steelblue')
axes[0].set_title('Số lượng mẫu theo ngành học')
axes[0].set_xlabel('Ngành')
axes[0].set_ylabel('Số lượng')
axes[0].tick_params(axis='x', rotation=45)

# Pie chart
counts.plot(kind='pie', ax=axes[1], autopct='%1.1f%%', startangle=90)
axes[1].set_title('Tỷ lệ phân phối ngành học')
axes[1].set_ylabel('')

plt.tight_layout()
plt.savefig('../data/samples/label_distribution.png', dpi=150)
plt.show()
print(f'Tổng số ngành: {df[label_col].nunique()}')
```

## 4. Phân phối điểm thi theo tổ hợp

```python
# TODO: thay danh sách môn thi thật
score_cols = ['toan', 'van', 'ly', 'hoa', 'sinh', 'anh', 'su', 'dia']
available_cols = [c for c in score_cols if c in df.columns]

if available_cols:
    df[available_cols].hist(bins=20, figsize=(15, 8), color='steelblue', edgecolor='white')
    plt.suptitle('Phân phối điểm thi từng môn', fontsize=14)
    plt.tight_layout()
    plt.savefig('../data/samples/score_distribution.png', dpi=150)
    plt.show()
else:
    print('⚠️  Chưa tìm thấy cột điểm — cập nhật score_cols cho đúng tên cột')
```

## 5. Correlation heatmap

```python
numeric_df = df.select_dtypes(include='number')
if not numeric_df.empty:
    plt.figure(figsize=(10, 8))
    sns.heatmap(numeric_df.corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
    plt.title('Heatmap tương quan các đặc trưng số')
    plt.tight_layout()
    plt.savefig('../data/samples/correlation_heatmap.png', dpi=150)
    plt.show()
else:
    print('⚠️  Không có cột số để vẽ heatmap')
```
